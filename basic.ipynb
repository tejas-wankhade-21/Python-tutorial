```python
#indexing in python is used to access the individual characters of a string. 
#The index of a string starts from 0. 
#For example, the index of the first character of a string is 0, the index of the second character is 1, and so on.
a = "production"
#    0123456789         indexing is starts from 0.

print(a[1:3]) #output: ro           #it gets the characters from index 1 to index 2. It does not include the character at index 3.

print(a[1:5]) #output: rodu         #it gets the characters from index 1 to index 4. It does not include the character at index 5.

print(a[2:]) #output: oduction     #it gets the characters from index 2 to the end of the string.
```

## Operators

```python
#operators in python are used to perform operations on variables and values.
#Arithmetic operators are used to perform mathematical operations on numeric values. 
# There are seven arithmetic operators in python:

#Addition (+): It is used to add two numbers. For example,
a = 5
b = 3
print(a + b) #output: 8          #it adds the two numbers and returns the result as an integer.


#Subtraction (-): It is used to subtract two numbers.
print(a - b) #output: 2          #it subtracts the two numbers and returns the result as an integer.


#Multiplication (*): It is used to multiply two numbers. 
print(a * b) #output: 15          #it multiplies the two numbers and returns the result as an integer.


#Division (/): It is used to divide two numbers.
print(a / b) #output: 1.6666666666666667    #it divides the two numbers and returns the result as a float.


#Modulus (%): It is used to find the remainder of a division operation.
print(a % b) #output: 2          #it finds the remainder of the division operation and returns the result as an integer.


#Exponentiation (**): It is used to raise a number to a power.
print(a ** b) #output: 125          #it raises the first number to the power of the second 
                                    #number and returns the result as an integer.


#Floor Division (//): It is used to divide two numbers and 
# return the largest integer less than or equal to the result.
d = 9
e = 5
print(d // e) #output: 2          #it divides the two numbers and returns the result as an integer.
```

### Exception handling

```python
# exception handling in python is a mechanism to handle runtime errors and maintain the normal flow of the program.

# try except else finally :-
# try except else finally are used to handle exceptions in python. The try block contains the code that may raise an exception, 
# the except block contains the code that will be executed if an exception occurs, 
# the else block contains the code that will be executed if no exception occurs, 
# and the finally block contains the code that will be executed regardless of whether an exception occurs or not.

try:
    a = int("abc")                                                          # it raises a ValueError because "abc" cannot be converted to an integer.
    print(a)                                                                # If you change the string to a valid integer, like "123", 
                                                                            # it will print 123 and the except block will not be executed.
    
except ValueError:
    print("ValueError: invalid literal for int() with base 10: 'abc'")      #it prints this line if a ValueError occurs in the try block.
    
else :
    print("No Error found")                                                 # it prints this line if no exception occurs in the try block.

finally:
    print("Execution completed")                                             # it prints this line regardless of whether an exception occurs or not.
```

```python
# You can create your own custom exceptions with using a raised exception. 
# This allows you to define specific error conditions for your application and handle them appropriately.

age = 23                                                # because age is a positive number, it can be considered valid. However, if age were negative, it would raise a ValueError.
if age < 0:                                             # age cannot be negative, so we raise a ValueError if the condition is met.
    raise ValueError("your age is in negative")
else:                                                   # if the age is not negative, we print a valid message. 
    print("your age is valid")
```

```python
# oops is used to create objects and classes in python. 
# It is a programming paradigm that uses objects and classes to organize code and data.
# it allows for encapsulation, inheritance, and polymorphism, which are key concepts in object-oriented programming.

class Car:                         # class is a blueprint for creating objects. It defines the properties and behaviors of the objects.
    def __init__(self):            # __init__ is a constructor method that is called when an object is created. It initializes the properties of the object.
        self.color = "Red"
        self.model = "TATA"
        self.tyres = 4
        self.warranty = "5 years"
        
        
nexon = Car()

print(nexon.color)
print(nexon.model)

print(nexon.tyres)
print(nexon.warranty)
```
